```python
dataset_root = "/home/geyuan/code/LIBERO-FT/libero/datasets/"
dataset_subname = "libero_90"
output_dir = os.path.join(dataset_root, "libero_force")
os.makedirs(output_dir, exist_ok=True)
video_save_dir = "/home/geyuan/code/LIBERO-FT/debug/replay_videos/"  # for visualization of replayed videos

## Modify the regex as you need to filter the hdf5 files to replay.
# pat = re.compile(r"(open|close).*drawer|microwave", re.IGNORECASE)
pat = re.compile(r"(STUDY_SCENE3).*left", re.IGNORECASE)

input_hdf5_fns = sorted(glob.glob(dataset_root + dataset_subname + "/*.hdf5"))
input_hdf5_fns = [fn for fn in input_hdf5_fns if pat.search(os.path.basename(fn))]
# print("Found hdf5 files:", input_hdf5_fns)
output_hdf5_fns = [
    os.path.join(
        output_dir,
        os.path.basename(fn).replace(".hdf5", "_wrench.hdf5")
    )
    for fn in input_hdf5_fns
]
print("Output hdf5 files:", output_hdf5_fns)


total_cnt = len(input_hdf5_fns)
for idx in tqdm(range(total_cnt), desc="Replaying"):
    input_hdf5_path = input_hdf5_fns[idx]
    output_hdf5_path = output_hdf5_fns[idx]

    replayer = HDF5Replayer(
        input_hdf5_path,
    )
    replayer.init_env()

    replayer.replay_hdf5_by_actions(
        hdf5_save_path=output_hdf5_path,
        video_save_dir=video_save_dir,
    )
    replayer.close_env()
```

## Start to replay with Visual shift

```python
OUT_DIR = "libero_force_view_slice"
DEMO_SLICES = "([0:50:3], [20:26])"
F_SCALE = 1.
G_SCALE = 1.
S_SCALE = 1.
DARKEN = False
COLOR_LIGHT = False
TEXTURE = False
VIEW_SHIFT = True
RE_PATTERN = r"(KITCHEN_SCENE1_open_the_top_drawer_of_the_cabinet_and_put_the_bowl)"
# RE_PATTERN = r"(STUDY_SCENE3).*left"

phy_keywords=(
    "cabinet", "drawer", "door", "basket", "object", "tool", "microwave",
    "gripper",
    "bowl",
    "book", "mug",
)

dataset_root = "/home/geyuan/code/LIBERO-FT/libero/datasets/"
dataset_subname = "libero_90"
output_dir = os.path.join(dataset_root, OUT_DIR, dataset_subname)
os.makedirs(output_dir, exist_ok=True)
video_save_dir = f"/home/geyuan/code/LIBERO-FT/debug/replay_videos/{OUT_DIR}"  # for visualization of replayed videos

## Modify the regex as you need to filter the hdf5 files to replay.
# pat = re.compile(r"(open|close).*drawer|microwave", re.IGNORECASE)
# pat = re.compile(r"(STUDY_SCENE3).*left", re.IGNORECASE)
pat = re.compile(RE_PATTERN, re.IGNORECASE)

input_hdf5_fns = sorted(glob.glob(dataset_root + dataset_subname + "/*.hdf5"))
input_hdf5_fns = [fn for fn in input_hdf5_fns if pat.search(os.path.basename(fn))]
print("Found hdf5 files:", input_hdf5_fns)
output_hdf5_fns = [
    os.path.join(
        output_dir,
        os.path.basename(fn).replace(".hdf5", "_wrench.hdf5")
    )
    for fn in input_hdf5_fns
]
print("Output hdf5 files:", output_hdf5_fns)


replay_start_seed = 1
total_cnt = len(input_hdf5_fns)
for idx in tqdm(range(total_cnt), desc="Replaying"):
    input_hdf5_path = input_hdf5_fns[idx]
    output_hdf5_path = output_hdf5_fns[idx]

    replayer = HDF5Replayer(
        input_hdf5_path,
        phy_keywords=phy_keywords,
    )
    replayer.init_env()

    stat = replayer.replay_hdf5_by_actions(
        hdf5_save_path=output_hdf5_path,
        video_save_dir=video_save_dir,
        save_max_videos=50,
        replay_seed=replay_start_seed,
        retry_with_reset=20,  # close the reset and retry mechanism
        friction_scale=F_SCALE,
        gz_scale=G_SCALE,
        solref_scale=S_SCALE,
        table_center_xyz=(-0.2, -0.2, 0.9),
        darken=DARKEN,
        color_light=COLOR_LIGHT,
        texture=TEXTURE,
        texture_dir="/home/geyuan/code/LIBERO-FT/images/dtd/",
        camera_shift=VIEW_SHIFT,
        demo_slices=DEMO_SLICES,
    )
    replay_start_seed += stat["total"]
    replayer.close_env()
```

## Start to replay with High Resolution

```python
OUT_DIR = "libero_force_1k"

dataset_root = "/home/geyuan/code/LIBERO-FT/libero/datasets/"
dataset_subname = "libero_10"  # `libero_90`, `libero_10`
output_dir = os.path.join(dataset_root, OUT_DIR, dataset_subname)
os.makedirs(output_dir, exist_ok=True)
video_save_dir = "/home/geyuan/code/LIBERO-FT/debug/replay_videos/"  # for visualization of replayed videos

## Modify the regex as you need to filter the hdf5 files to replay.
# pat = re.compile(r"(open|close).*drawer|microwave", re.IGNORECASE)
# pat = re.compile(r"(STUDY_SCENE3).*left", re.IGNORECASE)
pat = re.compile(r".*", re.IGNORECASE)

input_hdf5_fns = sorted(glob.glob(dataset_root + dataset_subname + "/*.hdf5"))
input_hdf5_fns = [fn for fn in input_hdf5_fns if pat.search(os.path.basename(fn))]
# print("Found hdf5 files:", input_hdf5_fns)
output_hdf5_fns = [
    os.path.join(
        output_dir,
        os.path.basename(fn).replace(".hdf5", "_wrench.hdf5")
    )
    for fn in input_hdf5_fns
]
print("Output hdf5 files:", output_hdf5_fns)


total_cnt = len(input_hdf5_fns)
for idx in tqdm(range(total_cnt), desc="Replaying"):
    input_hdf5_path = input_hdf5_fns[idx]
    output_hdf5_path = output_hdf5_fns[idx]

    replayer = HDF5Replayer(
        input_hdf5_path,
        camera_hw=(1024, 1024),
        retry_obs_source="env",
    )
    replayer.init_env()

    replayer.replay_hdf5_by_actions(
        hdf5_save_path=output_hdf5_path,
        video_save_dir=video_save_dir,
        save_max_videos=5,
    )
    replayer.close_env()
```
